```python
import hashlib


def _norm(x):
    if isinstance(x, bool):
        return repr(x)
    if isinstance(x, float):
        return f"{x:.2f}"
    if isinstance(x, int):
        return str(x)
    if isinstance(x, dict):
        return "{" + ",".join(f"{_norm(k)}:{_norm(v)}" for k, v in sorted(x.items(), key=lambda kv: _norm(kv[0]))) + "}"
    if isinstance(x, (set, frozenset)):
        return "{" + ",".join(sorted(_norm(v) for v in x)) + "}"
    if isinstance(x, (list, tuple)):
        return "[" + ",".join(_norm(v) for v in x) + "]"
    return repr(x)


ANSWERS = {
    "задача 1": "2c624232cd",
    "задача 2": "39188782c0",
    "задача 3": "0991f9b4ab",
    "задача 4": "5aad473f1f",
    "задача 5": "35f4c2e946",
    "задача 6": "82c3e9c7b3",
    "задача 7": "1819e16d75",
    "задача 8": "6c8dca6a70",
    "задача 9": "a6b948baf3",
    "задача 10": "690fd2761b",
    "задача 11": "4ca15026eb",
    "задача 12": "2cfb3e2d4e",
    "задача 13": "667ec9c99b",
    "задача 14": "c1b8543b7f"
}


def check(task: str, got) -> None:
    if got is None:
        print(f"{task}: не решено")
    elif hashlib.sha256(_norm(got).encode()).hexdigest()[:10] == ANSWERS.get(task):
        print(f"{task}: верно")
    else:
        print(f"{task}: не сходится, получилось {got!r}")


print("проверка готова, задач:", len(ANSWERS))
```
